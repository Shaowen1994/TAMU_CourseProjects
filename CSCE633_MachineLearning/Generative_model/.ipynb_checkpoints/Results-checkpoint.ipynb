```python
import numpy as np
import matplotlib.pyplot as plt
import math
```

#### Generator Loss

```python
gen_loss = np.loadtxt('Generator_loss')
x_range = range(1,1 + gen_loss.shape[0])
plt.plot(x_range,gen_loss)
plt.xlabel('Iteration')
plt.ylabel('Generator Loss')
plt.show()
```

#### Critic Loss

```python
cri_loss = np.loadtxt('Critic_loss')
x = [i/5 for i in range(cri_loss.shape[0])]
plt.plot(x,-cri_loss)
plt.xlabel('Iteration')
plt.ylabel('Critic Loss')
plt.show()
```

#### Some generated examples

```python
Sample_path = 'Generated_Samples/'

def load_samples(epoch):
    array = np.load(Sample_path + 'samples_{}.npy'.format(epoch))
    amount = array.shape[0]
    size_1 = math.ceil(math.sqrt(amount))
    size_2 = math.ceil(amount/float(size_1))
    plt.figure(figsize=(15,15))
    for i in range(amount):
        plt_index = i + 1
        plt.subplot(size_1, size_2, plt_index)
        plt.imshow(array[i])
        plt.axis('off')
    plt.show()
    return array
```

##### Epoch 1:

```python
array_1 = load_samples(1)
```
